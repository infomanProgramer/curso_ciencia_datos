## **Contenido**

- **Fundamentos de Ciencia de Datos:** Entender qué es, para qué sirve y cómo se trabaja con datos.
    - ¿Qué es la Ciencia de Datos?
    - Diferencias entre Data Science, Data Analytics y Business Intelligence.
    - Ciclo de vida de un proyecto de datos.
    - Tipos de datos: estructurados y no estructurados.
    - Fuentes de datos: Excel, CSV, bases de datos y APIs.
    - Casos de uso en empresas, comercios, educación y finanzas.
- **Python para Ciencia de Datos:** Variables, estructuras, funciones, colecciones y manejo de archivos.
    - Variables y tipos de datos.
    - Listas, tuplas, diccionarios y conjuntos.
    - Condicionales y ciclos.
    - Funciones.
    - Comprensión de listas.
    - Manejo de excepciones.
    - Lectura y escritura de archivos.
    - Introducción a Jupyter Notebook y Google Colab.
    - **Práctica:** crear un programa que lea ventas de un negocio y calcule totales, promedios y productos más vendidos.
- **Manipulación de datos con Pandas y NumPy:** Limpiar, transformar, filtrar y organizar información.
    - Series y DataFrames.
    - Importación de Excel y CSV.
    - Selección de filas y columnas.
    - loc e iloc.
    - Filtrado de registros.
    - Ordenamiento.
    - Agrupaciones con groupby().
    - Agregaciones: sum(), mean(), count().
    - Combinación de DataFrames: merge() y concat().
    - Tratamiento de valores nulos.
    - Duplicados y datos inconsistentes.
    - Conversión de tipos de datos.
    - Introducción a NumPy y operaciones vectorizadas.
    - **Práctica:** limpiar una base de datos de ventas que contiene valores nulos, duplicados, fechas incorrectas y montos en diferentes formatos.
- **Estadística descriptiva:** Medidas estadísticas, distribuciones, correlaciones y análisis.
    - Población y muestra.
    - Tipos de variables.
    - Media, mediana y moda.
    - Varianza y desviación estándar.
    - Cuartiles y percentiles.
    - Distribuciones de frecuencia.
    - Distribución normal.
    - Probabilidad básica.
    - Correlación.
    - Introducción a pruebas de hipótesis.
    - **Práctica:** analizar el comportamiento de las ventas de un negocio e identificar tendencias y diferencias entre períodos.
- **Visualización de datos:** Matplotlib, Seaborn y comunicación visual.
    - Principios de visualización.
    - Gráficos de barras.
    - Gráficos de líneas.
    - Histogramas.
    - Diagramas de dispersión.
    - Gráficos circulares.
    - Boxplots.
    - Gráficos combinados.
    - Personalización de gráficos.
    - Introducción a dashboards.
    - Interpretación y presentación de resultados.
    - **Práctica:** construir un informe visual de ventas con ingresos, egresos, utilidad, productos más vendidos y evolución mensual.
- **Bases de datos y SQL:** Consultas, filtros, agrupaciones y conexión con Python.
    - Fundamentos de bases de datos relacionales.
    - Tablas y relaciones.
    - SELECT, WHERE y ORDER BY.
    - GROUP BY y HAVING.
    - Funciones de agregación.
    - JOIN.
    - Subconsultas.
    - Conexión de Python con PostgreSQL.
    - Extracción de datos mediante consultas SQL.
    - Práctica: conectar Python a PostgreSQL y generar un análisis a partir de información almacenada en varias tablas.
- **Introducción al Machine Learning:** Modelos predictivos, entrenamiento, evaluación y clasificación.
    - ¿Qué es Machine Learning?
    - Aprendizaje supervisado y no supervisado.
    - Variables independientes y dependientes.
    - Conjuntos de entrenamiento y prueba.
    - Regresión lineal.
    - Regresión logística.
    - Árboles de decisión.
    - K-Means.
    - Métricas de evaluación.
    - Sobreajuste y subajuste.
    - Introducción a Scikit-learn.
    - Práctica: construir un modelo sencillo para predecir ventas o clasificar clientes.
- **Proyecto integrador:** Resolver un problema real utilizando todo lo aprendido.
    - Identificar un problema.
    - Obtener un conjunto de datos.
    - Explorar la información.
    - Limpiar y transformar los datos.
    - Aplicar análisis estadístico.
    - Construir visualizaciones.
    - Identificar patrones y tendencias.
    - Presentar conclusiones y recomendaciones.
